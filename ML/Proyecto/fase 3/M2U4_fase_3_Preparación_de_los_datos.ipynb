### **3.3. Tipos de variables**
---

La variable objetivo de este conjunto de datos es la de `Diabetes_012`, donde mediante las otras variables se busca determinar si un paciente va no tiene diabetes (0), es prediabético (1), o ya es diabético (2). Esta es una variable categórica que se representa de con un entero.

Para realizar el análisis descriptivo de la variable objetivo y de las otras, utilizaremos la siguiente función que obtiene las estadísticas principales de la variable.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def analisis_estadistico_variable(df, variable, log_scale=False):
    if variable not in df.columns:
        raise ValueError(f"La columna '{variable}' no existe.")

    data = df[variable].dropna()

    if not np.issubdtype(data.dtype, np.number):
        raise TypeError(f"La columna '{variable}' no es numérica.")

    if log_scale:
        if (data <= 0).any():
            data = data[data > 0]

    # --- Estadísticos ---
    mean = data.mean()
    median = data.median()
    mode = data.mode().iloc[0]
    std = data.std()
    min_val = data.min()
    max_val = data.max()
    kurtosis = data.kurtosis()
    skewness = data.skew()

    q1 = data.quantile(0.25)
    q2 = median
    q3 = data.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers_lower = data[data < lower_bound]
    porcentaje_lower = len(outliers_lower) / len(data) * 100
    outliers_upper = data[data > upper_bound]
    porcentaje_upper = len(outliers_upper) / len(data) * 100

    print("---------------------------------------------------------")
    print(f"\nVariable: {variable}")
    print(f"Escala: {'Logarítmica' if log_scale else 'Normal'}")
    print(f"Media: {mean:.4f}")
    print(f"Mediana (Q2): {q2:.4f}")
    print(f"Moda: {mode:.4f}")
    print(f"Desviación estándar: {std:.4f}")
    print(f"Curtosis: {kurtosis:.4f}")
    print(f"Asimetría: {skewness:.4f}")
    print("---------------------------------------------------------")
    print(f"Q1 (25%): {q1:.4f}")
    print(f"Q3 (75%): {q3:.4f}")
    print(f"IQR: {iqr:.4f}")
    print(f"Valor mínimo: {min_val:.4f}")
    print(f"Valor máximo: {max_val:.4f}")
    print(f"Rango: {max_val - min_val:.4f}")
    print(f"Valores atípicos inferiores: {len(outliers_lower)} ({porcentaje_lower:.4f}%)")
    print(f"Valores atípicos superiores: {len(outliers_upper)} ({porcentaje_upper:.4f}%)")
    print("---------------------------------------------------------")

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # ---------- HISTOGRAMA ----------
    sns.histplot(data, kde=True, ax=axes[0], color="#4C72B0", alpha=0.6)

    # ±1σ
    axes[0].axvspan(mean - std, mean + std, alpha=0.15, color="orange", label="±1σ")

    # Líneas principales
    axes[0].axvline(mean, color="red", linewidth=2, label="Media")
    axes[0].axvline(q2, color="black", linewidth=2, label="Mediana")
    axes[0].axvline(mode, color="purple", linestyle=":", linewidth=2, label="Moda")

    if log_scale:
        axes[0].set_xscale("log")

    axes[0].set_title(f"Distribución de {variable}")
    axes[0].set_xlabel(variable)
    axes[0].set_ylabel("Frecuencia")
    axes[0].legend()

    # ---------- BOXPLOT ----------
    sns.boxplot(x=data, ax=axes[1], color="#55A868")

    axes[1].axvline(mean, color="red", linewidth=2, label="Media")
    axes[1].axvspan(q1, q3, alpha=0.2, color="green", label="IQR")

    if log_scale:
        axes[1].set_xscale("log")

    axes[1].set_title(f"Boxplot de {variable}")
    axes[1].legend()

    plt.tight_layout()
    plt.show()
```

```python
analisis_estadistico_variable(diabetes_df, "Diabetes_012")
```

Como se puede observar, la variable objetivo está compuesta principalmente por la categoría 0, o no tiene diabetes. A esta le sigue la categoría de 2, tiene diabetes, y por último la categoría de prediabético. ---

El primer paso de limpieza consiste en verificar la existencia de valores faltantes y su posible distribución con respecto a la variable objetivo.

```python

missing_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros analizados",
            "Valores faltantes totales",
            "Filas con al menos un faltante",
            "Variables con faltantes",
        ],
        "Valor": [
            len(diabetes_df),
            total_nulls,
            int(diabetes_df.isna().any(axis=1).sum()),
            int((nulls_by_column > 0).sum()),
        ],
    }
)

missing_by_target = (
    diabetes_df.assign(Diabetes_012=diabetes_df["Diabetes_012"].astype(int))
    .groupby("Diabetes_012")
    .apply(lambda frame: int(frame.isna().sum().sum()))
    .rename("Valores faltantes")
    .reset_index()
)

display(missing_summary)
display(missing_by_target)
```

**Análisis y decisión sobre valores faltantes**

De acuerdo con los resultados anteriores, el dataset no presenta valores faltantes ni filas incompletas. Esto significa que la integridad del conjunto no se ve comprometida por ausencia de datos y, por tanto, no fue necesario aplicar estrategias de imputación, relleno o eliminación de registros.

También se verificó la distribución de faltantes respecto a la variable objetivo `Diabetes_012`, encontrando que ninguna de sus clases presenta vacíos de información. En consecuencia, la distribución original de la variable se mantiene intacta y no hay riesgo de introducir sesgos derivados de imputaciones artificiales en esta etapa.


### **4.2. Valores duplicados**
---

Posteriormente se evalúa la presencia de registros duplicados exactos. Dado que el dataset no contiene un identificador único de paciente, la igualdad total entre filas no permite confirmar identidad clínica individual; sin embargo, sí indica que el perfil observado ya está representado en el conjunto y puede sobreponderar artificialmente ciertos patrones si se conserva repetido.


```python
duplicate_mask = diabetes_df.duplicated(keep="first")
duplicate_profile_mask = diabetes_df.duplicated(keep=False)
duplicated_rows = int(duplicate_mask.sum())
rows_in_duplicate_profiles = int(duplicate_profile_mask.sum())

duplicate_profile_summary = (
    diabetes_df.groupby(diabetes_df.columns.tolist(), dropna=False, as_index=False)
    .size()
    .rename(columns={"size": "Frecuencia"})
    .query("Frecuencia > 1")
    .sort_values("Frecuencia", ascending=False)
    .reset_index(drop=True)
)

clean_df = diabetes_df.copy()

duplicate_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros iniciales",
            "Duplicados exactos adicionales",
            "Registros involucrados en perfiles duplicados",
            "Porcentaje de duplicados exactos adicionales",
            "Perfiles exactos únicos con frecuencia > 1",
            "Registros eliminados por deduplicación",
            "Registros conservados para el flujo principal",
        ],
        "Valor": [
            len(diabetes_df),
            duplicated_rows,
            rows_in_duplicate_profiles,
            round((duplicated_rows / len(diabetes_df)) * 100, 2),
            len(duplicate_profile_summary),
            0,
            len(clean_df),
        ],
    }
)

duplicate_class_distribution = (
    diabetes_df.loc[duplicate_profile_mask, "Diabetes_012"]
    .astype(int)
    .value_counts()
    .sort_index()
    .rename_axis("Diabetes_012")
    .reset_index(name="Registros en perfiles duplicados")
)

display(duplicate_summary)
display(duplicate_class_distribution)
display(duplicate_profile_summary.head())
```

**Análisis y decisión sobre duplicados**

La inspección confirma la existencia de perfiles repetidos exactamente iguales. En este proyecto estos duplicados **no se eliminan dentro del flujo supervisado principal**, porque se busca replicar la construcción del dataset binario balanceado reportada en la referencia metodológica sobre el conjunto original completo.

Aunque el dataset no incluye un identificador único de paciente, la presencia de perfiles repetidos se mantiene documentada como una posible fuente de sesgo y se conserva explícitamente para reproducir la estrategia de balanceo 50/50 previa al particionado. La deduplicación queda entonces como hallazgo analítico, pero no como una transformación aplicada al dataset base utilizado en la clasificación binaria.


### **4.3. Valores atípicos**
---

Con el diagnóstico de perfiles duplicados ya establecido, se revisan los valores atípicos en las variables numéricas más sensibles desde el punto de vista clínico.

```python
outlier_columns_clean = [col for col in ["BMI", "MentHlth", "PhysHlth"] if col in clean_df.columns]
outlier_rows_clean = []

for column in outlier_columns_clean:
    q1 = clean_df[column].quantile(0.25)
    q3 = clean_df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_count = int(((clean_df[column] < lower_bound) | (clean_df[column] > upper_bound)).sum())

    outlier_rows_clean.append(
        {
            "Variable": column,
            "Límite inferior IQR": round(lower_bound, 2),
            "Límite superior IQR": round(upper_bound, 2),
            "Posibles atípicos": outlier_count,
            "Porcentaje sobre clean_df": round((outlier_count / len(clean_df)) * 100, 2),
            "Mínimo observado": float(clean_df[column].min()),
            "Máximo observado": float(clean_df[column].max()),
        }
    )

clean_outlier_summary = pd.DataFrame(outlier_rows_clean)
display(clean_outlier_summary)
```

**Análisis y decisión sobre valores atípicos**

Los valores atípicos se identificaron mediante el criterio del rango intercuartílico (IQR), aplicado sobre `BMI`, `MentHlth` y `PhysHlth`, por ser variables donde los extremos pueden alterar la distribución de manera visible. Sin embargo, en este proyecto no se interpretan automáticamente como errores de medición.

Dado que se trata de un contexto clínico, los valores extremos pueden representar pacientes con condiciones de salud más severas, mayor carga de enfermedad o perfiles menos frecuentes pero reales. Por esta razón, la decisión final fue **conservar los valores atípicos** y controlar su efecto posteriormente mediante estandarización, en lugar de eliminarlos o transformarlos de forma agresiva.


### **4.4. Datos Inconsistentes**
---

En esta etapa se valida que las variables permanezcan dentro de los dominios esperados según la definición del dataset y se homogeneiza su representación para el modelado.

```python
binary_columns = [
    "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke", "HeartDiseaseorAttack",
    "PhysActivity", "Fruits", "Veggies", "HvyAlcoholConsump", "AnyHealthcare",
    "NoDocbcCost", "DiffWalk", "Sex"
]

domain_rules = {
    "Diabetes_012": {"allowed": [0, 1, 2]},
    "GenHlth": {"allowed": [1, 2, 3, 4, 5]},
    "Age": {"allowed": list(range(1, 14))},
    "Education": {"allowed": [1, 2, 3, 4, 5, 6]},
    "Income": {"allowed": list(range(1, 9))},
    "MentHlth": {"range": (0, 30)},
    "PhysHlth": {"range": (0, 30)},
    "BMI": {"range": (1, None)},
}

for column in binary_columns:
    domain_rules[column] = {"allowed": [0, 1]}

inconsistent_row_mask = pd.Series(False, index=clean_df.index)
domain_check_rows = []

for column, rule in domain_rules.items():
    series = clean_df[column]
    if "allowed" in rule:
        invalid_mask = ~series.isin(rule["allowed"])
        criterio = f"Valores permitidos: {rule['allowed']}"
    else:
        lower, upper = rule["range"]
        invalid_mask = pd.Series(False, index=series.index)
        if lower is not None:
            invalid_mask |= series < lower
        if upper is not None:
            invalid_mask |= series > upper
        criterio = f"Rango esperado: {lower} a {upper if upper is not None else 'inf'}"

    inconsistent_row_mask |= invalid_mask
    domain_check_rows.append(
        {
            "Variable": column,
            "Criterio": criterio,
            "Registros fuera de dominio": int(invalid_mask.sum()),
        }
    )

domain_check_summary = pd.DataFrame(domain_check_rows)
integer_like_summary = pd.DataFrame(
    {
        "Variable": clean_df.columns,
        "Valores enteros codificados": [bool(np.allclose(clean_df[col], clean_df[col].round())) for col in clean_df.columns],
    }
)

clean_df = clean_df.round().astype(int)
consistent_df = clean_df.loc[~inconsistent_row_mask].reset_index(drop=True).copy()

display(domain_check_summary)
display(integer_like_summary)
display(pd.DataFrame({
    "Indicador": ["Registros con inconsistencias", "Registros después de validación de dominios"],
    "Valor": [int(inconsistent_row_mask.sum()), len(consistent_df)]
}))
```

**Análisis y decisión sobre inconsistencias**

La validación se realizó contrastando cada variable con su dominio esperado. En las variables binarias se comprobó la presencia exclusiva de valores `0` y `1`; en la variable objetivo se verificó el conjunto `{0, 1, 2}`; y en las variables ordinales y de conteo se revisaron sus rangos clínicamente plausibles.

Además, se observó que el archivo fue importado como `float64`, aunque sus valores están codificados como enteros. Para homogeneizar el formato y mejorar la coherencia del análisis, se convirtieron estas columnas a enteros antes de la fase de modelado. Esta decisión no altera la información, pero sí deja el dataset con una representación más consistente respecto a su significado original.


### **4.5. Datos corruptos**
---

Como complemento de la validación anterior, se revisa si existen registros corruptos derivados de errores de lectura, coerción numérica o valores infinitos.

```python
coerced_df = consistent_df.apply(pd.to_numeric, errors="coerce")
coercion_failures = int(((coerced_df.isna()) & (~consistent_df.isna())).sum().sum())
infinite_values = int(np.isinf(coerced_df.to_numpy()).sum())
corrupt_row_mask = coerced_df.isna().any(axis=1) | np.isinf(coerced_df.to_numpy()).any(axis=1)

corrupt_summary = pd.DataFrame(
    {
        "Indicador": [
            "Fallos de coerción numérica",
            "Valores infinitos",
            "Registros corruptos detectados",
        ],
        "Valor": [
            coercion_failures,
            infinite_values,
            int(corrupt_row_mask.sum()),
        ],
    }
)

prepared_df = consistent_df.loc[~corrupt_row_mask].reset_index(drop=True).copy()

display(corrupt_summary)
display(pd.DataFrame({
    "Indicador": ["Registros listos después del control de corrupción"],
    "Valor": [len(prepared_df)]
}))
```

**Análisis y decisión sobre datos corruptos**

La búsqueda de datos corruptos se enfocó en tres señales: valores no convertibles a formato numérico, valores infinitos y registros que, después de la validación de dominios, siguieran mostrando comportamientos incompatibles con la estructura del dataset. Esta revisión es especialmente importante antes de alimentar algoritmos como una red neuronal o `K-means`, que requieren matrices numéricas consistentes.

Con base en la inspección realizada, el conjunto no presenta evidencia de corrupción estructural. Por ello, no fue necesario eliminar registros adicionales por este motivo y el dataset preparado conserva las observaciones válidas y consistentes del conjunto original tras la validación de dominios y el control de corrupción.


### **4.6. Selección de datos**
---

Finalmente, se define qué variables se conservan para cada uno de los enfoques de modelado y se justifica su inclusión.

```python
target_column = "Diabetes_012"
supervised_target_column = "Diabetes_binary"
supervised_target_mapping = {0: 0, 1: 1, 2: 1}
class_ids = [0, 1]
class_labels = ["Sin diabetes (0)", "Prediabetes/Diabetes (1)"]

prepared_df[supervised_target_column] = (
    prepared_df[target_column].replace(supervised_target_mapping).astype(int)
)

feature_columns = [
    column for column in prepared_df.columns
    if column not in [target_column, supervised_target_column]
]

binary_feature_columns = [column for column in binary_columns if column in feature_columns]
ordinal_feature_columns = [column for column in ["GenHlth", "Age", "Education", "Income"] if column in feature_columns]
count_feature_columns = [column for column in ["BMI", "MentHlth", "PhysHlth"] if column in feature_columns]

selection_summary = pd.DataFrame(
    [
        {"Grupo": "Objetivo original", "Variables": target_column, "Uso supervisado": "Referencia clínica original", "Uso K-means": "Se excluye"},
        {"Grupo": "Objetivo supervisado", "Variables": supervised_target_column, "Uso supervisado": "Variable objetivo binaria", "Uso K-means": "Se excluye"},
        {"Grupo": "Binarias", "Variables": ", ".join(binary_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
        {"Grupo": "Ordinales", "Variables": ", ".join(ordinal_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
        {"Grupo": "Conteo / magnitud", "Variables": ", ".join(count_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
    ]
)

selection_size_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros disponibles para modelado",
            "Variables predictoras",
            "Variables en el problema supervisado",
            "Clases del problema supervisado",
            "Variables en el problema no supervisado",
        ],
        "Valor": [
            len(prepared_df),
            len(feature_columns),
            len(feature_columns) + 1,
            len(class_ids),
            len(feature_columns),
        ],
    }
)

display(selection_summary)
display(selection_size_summary)
```

**Análisis y decisión sobre selección de datos**

No se aplica reducción del tamaño del conjunto en la preparación base, ya que para el flujo supervisado se busca partir del dataset original validado y luego construir una versión binaria balanceada 50/50. Las demás variables se mantienen como predictoras porque aportan información sobre condiciones médicas, hábitos de vida y contexto sociodemográfico.

Para el enfoque supervisado se conserva `Diabetes_012` como referencia clínica original y se construye `Diabetes_binary`, donde `0` representa ausencia de diabetes y `1` agrupa los casos de prediabetes y diabetes. A partir de esta variable se construye después un dataset balanceado 50/50 usando todos los casos positivos y una muestra aleatoria de no diabéticos del mismo tamaño. Para el enfoque no supervisado con `K-means`, se excluyen ambas etiquetas y se utiliza únicamente el espacio de características, con el fin de explorar agrupamientos sin incorporar la variable clínica objetivo en el proceso de clustering.


## **5. Dataset listo para modelado**
---

A continuación se generan dos salidas listas para la siguiente fase: una para clasificación supervisada binaria con `Diabetes_binary` y otra para `K-means`.


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

positive_risk_df = prepared_df[
    prepared_df[supervised_target_column] == 1
].copy()
negative_df = prepared_df[
    prepared_df[supervised_target_column] == 0
].copy()

negative_sample_df = negative_df.sample(
    n=len(positive_risk_df),
    replace=False,
    random_state=100,
)

balanced_supervised_df = (
    pd.concat([negative_sample_df, positive_risk_df], axis=0)
    .sample(frac=1, random_state=100)
    .reset_index(drop=True)
)

X_supervised = balanced_supervised_df[feature_columns].copy()
y_supervised = balanced_supervised_df[supervised_target_column].astype(int).copy()

X_train, X_test, y_train, y_test = train_test_split(
    X_supervised,
    y_supervised,
    test_size=0.2,
    random_state=42,
    stratify=y_supervised,
)

supervised_scaler = StandardScaler()
X_train_scaled = pd.DataFrame(
    supervised_scaler.fit_transform(X_train),
    columns=feature_columns,
    index=X_train.index,
)
X_test_scaled = pd.DataFrame(
    supervised_scaler.transform(X_test),
    columns=feature_columns,
    index=X_test.index,
)

X_unsupervised = prepared_df[feature_columns].copy()
kmeans_scaler = StandardScaler()
X_kmeans_scaled = pd.DataFrame(
    kmeans_scaler.fit_transform(X_unsupervised),
    columns=feature_columns,
    index=X_unsupervised.index,
)

binary_distribution_summary = pd.concat(
    [
        prepared_df[supervised_target_column].value_counts().sort_index().rename("Original"),
        balanced_supervised_df[supervised_target_column].value_counts().sort_index().rename("Balanceado 50/50"),
        y_train.value_counts().sort_index().rename("Train"),
        y_test.value_counts().sort_index().rename("Test"),
    ],
    axis=1,
).fillna(0).astype(int)
binary_distribution_summary.index.name = supervised_target_column
binary_distribution_summary["Clase"] = binary_distribution_summary.index.map(
    dict(zip(class_ids, class_labels))
)
binary_distribution_summary = binary_distribution_summary.reset_index()[
    [supervised_target_column, "Clase", "Original", "Balanceado 50/50", "Train", "Test"]
]

split_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros originales supervisados",
            "Registros en el dataset binario balanceado",
            "Registros en train",
            "Registros en test",
            "Proporción train sobre dataset balanceado",
            "Proporción test sobre dataset balanceado",
        ],
        "Valor": [
            len(prepared_df),
            len(balanced_supervised_df),
            len(X_train),
            len(X_test),
            round(len(X_train) / len(balanced_supervised_df), 2),
            round(len(X_test) / len(balanced_supervised_df), 2),
        ],
    }
)

model_ready_summary = pd.DataFrame(
    {
        "Dataset": [
            "Supervisado balanceado - X_train_scaled",
            "Supervisado balanceado - X_test_scaled",
            "Supervisado balanceado - y_train",
            "Supervisado balanceado - y_test",
            "No supervisado - X_kmeans_scaled",
        ],
        "Filas": [
            len(X_train_scaled),
            len(X_test_scaled),
            len(y_train),
            len(y_test),
            len(X_kmeans_scaled),
        ],
        "Columnas": [
            X_train_scaled.shape[1],
            X_test_scaled.shape[1],
            1,
            1,
            X_kmeans_scaled.shape[1],
        ],
    }
)

display(split_summary)
display(binary_distribution_summary)
display(model_ready_summary)
display(balanced_supervised_df.head())
display(X_train_scaled.head())
display(X_kmeans_scaled.head())
```

**Resultado final de la preparación**

Después del proceso de limpieza, el conjunto queda listo para los dos enfoques definidos. En el caso supervisado, `Diabetes_012` se transforma en `Diabetes_binary`, donde la clase `0` representa ausencia de diabetes y la clase `1` agrupa prediabetes y diabetes. Sobre esa variable se construye primero un dataset balanceado 50/50, usando todos los individuos en riesgo y una muestra aleatoria de no diabéticos del mismo tamaño. Luego se realiza el `train/test split` estratificado sobre ese dataset ya balanceado.

Posteriormente, las variables predictoras se escalan para facilitar el entrenamiento de los modelos de clasificación. Para el enfoque no supervisado, se prepara una matriz `X_kmeans_scaled` sin las etiquetas objetivo y con todas las características estandarizadas, de modo que `K-means` opere sobre variables comparables en escala. En consecuencia, el flujo supervisado principal ya no requiere una técnica adicional de balanceo después de la partición.


```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Definimos los modelos a evaluar
modelos = {
    "Regresión Logística": LogisticRegression(max_iter=1000, random_state=100),
    "Árbol de Decisión": DecisionTreeClassifier(max_depth=10, random_state=100),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=10, random_state=100)
}

predicciones = {}

# Entrenamos iterativamente cada modelo
for nombre, modelo in modelos.items():
    modelo.fit(X_train_scaled, y_train)
    predicciones[nombre] = modelo.predict(X_test_scaled)
```

```python
for nombre, y_pred in predicciones.items():
    print("-" * 50)
    print(f"--- Resultados para: {nombre} ---")
    print("-" * 50)
    print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print()
    print(classification_report(
        y_test,
        y_pred,
        labels=class_ids,
        target_names=class_labels,
    ))

    cm = confusion_matrix(y_test, y_pred, labels=class_ids)
    plt.figure(figsize=(7, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        xticklabels=class_labels,
        yticklabels=class_labels,
    )
    plt.title(f'Matriz de Confusión - {nombre}')
    plt.ylabel('Valor Real')
    plt.xlabel('Predicción')
    plt.show()
```

```python
import pandas as pd

rf_modelo = modelos["Random Forest"]

importancias = rf_modelo.feature_importances_
columnas = X_train_scaled.columns

importancia_df = pd.DataFrame({'Variable': columnas, 'Importancia': importancias})
importancia_df = importancia_df.sort_values(by='Importancia', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x='Importancia', y='Variable', data=importancia_df, palette='viridis')
plt.title('Importancia de las Variables - Random Forest')
plt.xlabel('Nivel de importancia')
plt.ylabel('Variable')
plt.show()
```

```python
from sklearn.feature_selection import SelectFromModel
from sklearn.metrics import classification_report

print("--- SELECCIONANDO LAS MEJORES VARIABLES ---")
selector = SelectFromModel(modelos["Random Forest"], prefit=True, threshold='mean')

X_train_reducido = selector.transform(X_train_scaled)
X_test_reducido = selector.transform(X_test_scaled)

variables_elegidas = X_train_scaled.columns[selector.get_support()]
print(f"De {X_train_scaled.shape[1]} variables originales, nos quedamos con {len(variables_elegidas)}:")
print(variables_elegidas.tolist())
print()

predicciones_reducidas = {}

for nombre, modelo in modelos.items():
    modelo.fit(X_train_reducido, y_train)
    y_pred_reducido = modelo.predict(X_test_reducido)
    predicciones_reducidas[nombre] = y_pred_reducido

    print()
    print(f"--- Resultados para: {nombre} (Datos Reducidos) ---")
    print(f"Accuracy: {accuracy_score(y_test, y_pred_reducido):.4f}")
    print(classification_report(
        y_test,
        y_pred_reducido,
        labels=class_ids,
        target_names=class_labels,
    ))
    print("-" * 50)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

for nombre, y_pred in predicciones_reducidas.items():
    cm = confusion_matrix(y_test, y_pred, labels=class_ids)

    plt.figure(figsize=(7, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        xticklabels=class_labels,
        yticklabels=class_labels,
    )

    plt.title(f'Matriz de Confusión (Datos Reducidos) - {nombre}', pad=15)
    plt.ylabel('Valor Real', fontweight='bold')
    plt.xlabel('Predicción', fontweight='bold')
    plt.show()
```

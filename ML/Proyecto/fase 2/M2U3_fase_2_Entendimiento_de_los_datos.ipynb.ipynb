#**Origen**

Los datos utilizados en este proyecto provienen del sistema de encuestas de salud conocido como Behavioral Risk Factor Surveillance System (BRFSS), desarrollado por el Centers for Disease Control and Prevention.

El BRFSS es una encuesta telefónica anual que recopila información sobre comportamientos de riesgo en salud, condiciones crónicas y uso de servicios preventivos en la población adulta de Estados Unidos. Para este proyecto se utiliza una versión procesada y limpia del dataset correspondiente al año 2015, disponible en la plataforma
Kaggle.

El conjunto de datos contiene información de más de 250,000 individuos y un conjunto de aproximadamente 21 variables relacionadas con indicadores de salud, hábitos de vida y condiciones médicas. Estas variables son principalmente de tipo numérico o categórico.

Adicionalmente, el dataset [Diabetes Health Indicators Dataset](https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset/data) se presenta en tres versiones, la primera es donde tenemos pacientes diabéticos, prediabeticos y los no diabéticos. La segunda versión es una balanceada  es decir la mitad de los registros tienen diabetes y los demás no la presentan y una tercera versión binaria no balanceada, que refleja la distribución real de la población.

## Descripción de Variables del Dataset

Describimos las variables del dataset donde las clasificamos de acuerdo a la información que nos aporta

- **Diabetes_012:** Variable objetivo del modelo. Clasifica a los individuos en tres categorías
  - 0: No presenta diabetes o solo durante el embarazo  
  - 1: Prediabetes  
  - 2: Diabetes  


###  Variables de condiciones médicas

- **HighBP:** Indica si el individuo presenta hipertensión (presión arterial alta).  
- **HighChol:** Indica si el individuo tiene niveles altos de colesterol.  
- **CholCheck:** Señala si la persona se ha realizado un chequeo de colesterol en los últimos años.  
- **Stroke:** Indica si el individuo ha sufrido un accidente cerebrovascular.  
- **HeartDiseaseorAttack:** Indica si ha tenido enfermedad cardíaca o ataque al corazón.  



###  Variables de hábitos y estilo de vida

- **BMI:** Índice de masa corporal del individuo, relacionado con el sobrepeso y la obesidad.  
- **Smoker:** Indica si la persona es fumadora.  
- **PhysActivity:** Señala si el individuo realiza actividad física regularmente.  
- **Fruits:** Indica consumo frecuente de frutas.  
- **Veggies:** Indica consumo frecuente de verduras.  
- **HvyAlcoholConsump:** Indica consumo elevado de alcohol.  



###  Variables de acceso al sistema de salud

- **AnyHealthcare:** Indica si el individuo cuenta con algún tipo de cobertura médica.  
- **NoDocbcCost:** Indica si la persona no pudo acudir al médico debido a costos.  



###  Variables de estado de salud general

- **GenHlth:** Percepción general del estado de salud (escala ordinal).  
- **MentHlth:** Número de días con problemas de salud mental en un período determinado.  
- **PhysHlth:** Número de días con problemas de salud física.  
- **DiffWalk:** Indica si el individuo presenta dificultades para caminar o moverse.  



###  Variables sociodemográficas

- **Sex:** Sexo del individuo.  
- **Age:** Edad (generalmente codificada en rangos).  
- **Education:** Nivel educativo alcanzado.  
- **Income:** Nivel de ingresos.

### **3.2. Resumen de la Calidad de los datos**
---

En esta sección se revisan faltantes, duplicados, consistencia de tipos de datos y posibles valores atípicos en variables numéricas relevantes.

```python
nulls_by_column = diabetes_df.isna().sum().sort_values(ascending=False)
total_nulls = int(nulls_by_column.sum())
fully_empty_rows = int(diabetes_df.isna().all(axis=1).sum())
duplicated_rows = int(diabetes_df.duplicated().sum())
object_columns = diabetes_df.select_dtypes(include="object").columns.tolist()
blank_string_cells = 0

if object_columns:
    blank_string_cells = int(
        diabetes_df[object_columns]
        .apply(lambda col: col.astype(str).str.strip().eq("").sum())
        .sum()
    )

quality_summary = pd.DataFrame(
    {
        "Indicador": [
            "Valores nulos totales",
            "Filas completamente vacías",
            "Registros duplicados exactos",
            "Columnas tipo texto",
            "Celdas vacías en columnas tipo texto",
        ],
        "Valor": [
            total_nulls,
            fully_empty_rows,
            duplicated_rows,
            len(object_columns),
            blank_string_cells,
        ],
    }
)

quality_summary
```

```python
pd.DataFrame({"Variable": nulls_by_column.index, "Valores nulos": nulls_by_column.values})
```

```python
dtype_summary = (
    diabetes_df.dtypes.astype(str)
    .reset_index()
    .rename(columns={"index": "Variable", 0: "Tipo de dato"})
)

dtype_summary
```

```python
unique_summary = (
    diabetes_df.nunique()
    .sort_values()
    .reset_index()
    .rename(columns={"index": "Variable", 0: "Valores únicos"})
)

unique_summary
```

```python
diabetes_df.describe().T
```

```python
outlier_columns = [col for col in ["BMI", "MentHlth", "PhysHlth"] if col in diabetes_df.columns]
outlier_rows = []

for column in outlier_columns:
    q1 = diabetes_df[column].quantile(0.25)
    q3 = diabetes_df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_count = int(((diabetes_df[column] < lower_bound) | (diabetes_df[column] > upper_bound)).sum())

    outlier_rows.append(
        {
            "Variable": column,
            "Límite inferior IQR": round(lower_bound, 2),
            "Límite superior IQR": round(upper_bound, 2),
            "Posibles atípicos": outlier_count,
            "Mínimo observado": float(diabetes_df[column].min()),
            "Máximo observado": float(diabetes_df[column].max()),
        }
    )

outlier_summary = pd.DataFrame(outlier_rows)
outlier_summary
```

**Resumen de la Calidad de los datos**

Con base en las tablas anteriores, la revisión inicial permite verificar si existen valores nulos, filas completamente vacías y registros duplicados exactos dentro del conjunto. Esta inspección es importante porque cualquier ausencia de datos, duplicado o inconsistencia debe considerarse antes de pasar a la etapa de modelado.

En términos de estructura, el dataset presenta un formato homogéneo y tabular. Además, la revisión de tipos de dato permite confirmar que no hay problemas evidentes de codificación ni documentos ilegibles, ya que las variables se interpretan de forma consistente para su análisis en `pandas`.

También se incorporó una revisión descriptiva y una detección preliminar de posibles valores atípicos en variables como `BMI`, `MentHlth` y `PhysHlth`. En este proyecto, dichos valores atípicos no se eliminarán automáticamente, ya que en un contexto clínico pueden corresponder a casos reales y clínicamente relevantes.

En consecuencia, la identificación mediante el criterio del rango intercuartílico se utiliza únicamente como apoyo exploratorio para reconocer observaciones extremas, pero no como una regla de depuración. Conservar estos registros reduce el riesgo de perder información valiosa sobre pacientes con condiciones de salud más severas o poco frecuentes.

La variable objetivo de este conjunto de datos es `Diabetes_binary`, donde mediante las otras variables se busca determinar si un paciente no tiene diabetes (`0`) o si presenta prediabetes o diabetes (`1`). Esta es una variable categórica binaria representada mediante enteros.

Para realizar el análisis descriptivo de la variable objetivo y de las otras, utilizaremos la siguiente función que obtiene las estadísticas principales de la variable.


```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def analisis_estadistico_variable(df, variable, log_scale=False):
    if variable not in df.columns:
        raise ValueError(f"La columna '{variable}' no existe.")

    data = df[variable].dropna()

    if not np.issubdtype(data.dtype, np.number):
        raise TypeError(f"La columna '{variable}' no es numérica.")

    if log_scale:
        if (data <= 0).any():
            data = data[data > 0]

    # --- Estadísticos ---
    mean = data.mean()
    median = data.median()
    mode = data.mode().iloc[0]
    std = data.std()
    min_val = data.min()
    max_val = data.max()
    kurtosis = data.kurtosis()
    skewness = data.skew()

    q1 = data.quantile(0.25)
    q2 = median
    q3 = data.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers_lower = data[data < lower_bound]
    porcentaje_lower = len(outliers_lower) / len(data) * 100
    outliers_upper = data[data > upper_bound]
    porcentaje_upper = len(outliers_upper) / len(data) * 100

    print("---------------------------------------------------------")
    print(f"\nVariable: {variable}")
    print(f"Escala: {'Logarítmica' if log_scale else 'Normal'}")
    print(f"Media: {mean:.4f}")
    print(f"Mediana (Q2): {q2:.4f}")
    print(f"Moda: {mode:.4f}")
    print(f"Desviación estándar: {std:.4f}")
    print(f"Curtosis: {kurtosis:.4f}")
    print(f"Asimetría: {skewness:.4f}")
    print("---------------------------------------------------------")
    print(f"Q1 (25%): {q1:.4f}")
    print(f"Q3 (75%): {q3:.4f}")
    print(f"IQR: {iqr:.4f}")
    print(f"Valor mínimo: {min_val:.4f}")
    print(f"Valor máximo: {max_val:.4f}")
    print(f"Rango: {max_val - min_val:.4f}")
    print(f"Valores atípicos inferiores: {len(outliers_lower)} ({porcentaje_lower:.4f}%)")
    print(f"Valores atípicos superiores: {len(outliers_upper)} ({porcentaje_upper:.4f}%)")
    print("---------------------------------------------------------")

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # ---------- HISTOGRAMA ----------
    sns.histplot(data, kde=True, ax=axes[0], color="#4C72B0", alpha=0.6)

    # ±1σ
    axes[0].axvspan(mean - std, mean + std, alpha=0.15, color="orange", label="±1σ")

    # Líneas principales
    axes[0].axvline(mean, color="red", linewidth=2, label="Media")
    axes[0].axvline(q2, color="black", linewidth=2, label="Mediana")
    axes[0].axvline(mode, color="purple", linestyle=":", linewidth=2, label="Moda")

    if log_scale:
        axes[0].set_xscale("log")

    axes[0].set_title(f"Distribución de {variable}")
    axes[0].set_xlabel(variable)
    axes[0].set_ylabel("Frecuencia")
    axes[0].legend()

    # ---------- BOXPLOT ----------
    sns.boxplot(x=data, ax=axes[1], color="#55A868")

    axes[1].axvline(mean, color="red", linewidth=2, label="Media")
    axes[1].axvspan(q1, q3, alpha=0.2, color="green", label="IQR")

    if log_scale:
        axes[1].set_xscale("log")

    axes[1].set_title(f"Boxplot de {variable}")
    axes[1].legend()

    plt.tight_layout()
    plt.show()
```

### **3.4. Relación Entre Variables**
---

Para revisar las posibles relaciones entre las variables, emplearemos la función de `sklearn` `corr` para obtener una matriz de correlaciones y así determinar qué variables están altamente correlaciones. Emplearemos un mapa de calor para facilitar la visualización.

```
corr_matrix = diabetes_df.corr()
```

```
def top_correlated_features(corr_matrix, target_var, top_n=10, use_absolute=True):
    if target_var not in corr_matrix.columns:
        raise ValueError(f"{target_var} no está en la matriz de correlación")

    # Extraer correlaciones con la variable objetivo
    corr_series = corr_matrix[target_var].drop(labels=[target_var])

    if use_absolute:
        corr_series = corr_series.reindex(corr_series.abs().sort_values(ascending=False).index)
    else:
        corr_series = corr_series.sort_values(ascending=False)

    # Seleccionar top N
    top_corr = corr_series.head(top_n).reset_index()
    top_corr.columns = ['Variable', 'Correlación']

    return top_corr
```

```
top_corr = top_correlated_features(corr_matrix, target_var='Diabetes_binary', top_n=10)
print(top_corr)
```

La celda anterior nos trae las variables con mayor correlación a la variable objetivo de Diabete_binary. En general, notamos que hay una correlación positiva, aunque no muy fuerte, con las variables de salud. Por ejemplo, GenHlth tiene la escala de que un valor de 1 indica excelente salud mientras que 5 indica pésima salud, tiene sentido que exista la correlación positiva, ya que el valor de 1 en Diabetes indica la presencia de la enfermedad. Así mismo, HighBP, HighChol y las demás variables dicotomizadas presentan la misma tendencia. Por otro lado, las escalas de Income y Education son más tradicionales, con valores elevados indicando mejores condiciones. Su correlación con la presencia de diabetes es negativa, aunque bastante debil.

```
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)
corr_pairs = corr_matrix.where(mask)

corr_pairs = corr_pairs.stack().reset_index()
corr_pairs.columns = ['Variable 1', 'Variable 2', 'Correlación']

corr_pairs['AbsCorrelation'] = corr_pairs['Correlación'].abs()
corr_pairs = corr_pairs.sort_values(by='AbsCorrelation', ascending=False)

top_n = 10
top_corr = corr_pairs.head(top_n)

print(top_corr)
```

De la celda anterior, obtenemos los diez pares de variables más correlacionados. Por lo general, se ve una tendencia en que las variables de salud estén correlacionadas entre sí, al igual que algunas sociodemográficas. En particular, en este top diez aparece que nuestra variable objetivo tiene la mayor correlación con la variable de `GenHlth`, o la percepción general de la salud. Sin embargo, los coeficientes de correlación están enter 0.52 y 0.0, lo cual nos podría indicar que en general hay una correlación moderada y debil a lo largo del conjunto de datos. Esto es algo favorable ya que significa que el conjunto de datos no presente correlaciones lineales fuertes y por lo tanto todas las variables puede aportar buena varianza.

```
from scipy.stats import chi2_contingency

cramers_binary_columns = [
    "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke", "HeartDiseaseorAttack",
    "PhysActivity", "Fruits", "Veggies", "HvyAlcoholConsump", "AnyHealthcare",
    "NoDocbcCost", "DiffWalk", "Sex"
]

cramers_ordinal_columns = ["GenHlth", "Age", "Education", "Income"]
target_column = "Diabetes_binary"
cramers_columns = [
    column for column in (cramers_binary_columns + cramers_ordinal_columns)
    if column in diabetes_df.columns
]

def cramers_v(x, y):
    contingency_table = pd.crosstab(x, y)
    chi2, p_value, dof, _ = chi2_contingency(contingency_table, correction=False)
    n = contingency_table.to_numpy().sum()
    min_dim = min(contingency_table.shape) - 1

    if n == 0 or min_dim <= 0:
        return np.nan, chi2, p_value, dof

    v = np.sqrt(chi2 / (n * min_dim))
    return v, chi2, p_value, dof

cramers_rows = []

for column in cramers_columns:
    v_value, chi2, p_value, dof = cramers_v(diabetes_df[target_column], diabetes_df[column])
    cramers_rows.append(
        {
            "Variable": column,
            "V de Cramer": v_value,
            "Chi-cuadrado": chi2,
            "p-valor": p_value,
            "Grados de libertad": dof,
        }
    )

cramers_summary = (
    pd.DataFrame(cramers_rows)
    .sort_values(by="V de Cramer", ascending=False)
    .reset_index(drop=True)
)

display(cramers_summary.round({"V de Cramer": 4, "Chi-cuadrado": 2, "p-valor": 6}))
```

**Lectura del resultado con V de Cramer**

La tabla anterior complementa la correlación lineal de Pearson con una medida más adecuada para variables categóricas. En este caso, `V de Cramer` cuantifica la fuerza de asociación entre `Diabetes_binary` y cada predictor binario u ordinal, por lo que las variables ubicadas en las primeras posiciones del ranking son las que muestran una relación categórica más fuerte con la presencia de diabetes.

Este resultado debe interpretarse como evidencia de asociación y no de causalidad. Además, permite contrastar si las variables que aparecían como más relevantes en la matriz de correlación también conservan una señal alta cuando se analizan desde tablas de contingencia, especialmente en variables clínicas y de percepción general de salud.


## **4. Limpieza de los Datos**
---

En esta sección se realiza la preparación del conjunto de datos para que pueda ser utilizado en dos enfoques posteriores: un modelo supervisado de clasificación binaria, donde se fusionan los casos de prediabetes y diabetes en una sola clase positiva, y un modelo no supervisado con `K-means`.

Las decisiones de limpieza se toman con base en dos criterios principales. El primero es preservar la validez clínica de la información, evitando eliminar automáticamente registros que puedan representar casos reales de mayor riesgo. El segundo es reducir sesgos innecesarios en el modelado, especialmente aquellos introducidos por duplicados exactos, errores de formato o problemas de consistencia que sí afectan la estabilidad de los algoritmos.


### **4.1. Valores faltantes**
---

El primer paso de limpieza consiste en verificar la existencia de valores faltantes y su posible distribución con respecto a la variable objetivo.

```
missing_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros analizados",
            "Valores faltantes totales",
            "Filas con al menos un faltante",
            "Variables con faltantes",
        ],
        "Valor": [
            len(diabetes_df),
            total_nulls,
            int(diabetes_df.isna().any(axis=1).sum()),
            int((nulls_by_column > 0).sum()),
        ],
    }
)

missing_by_target = (
    diabetes_df.assign(Diabetes_binary=diabetes_df["Diabetes_binary"].astype(int))
    .groupby("Diabetes_binary")
    .apply(lambda frame: int(frame.isna().sum().sum()))
    .rename("Valores faltantes")
    .reset_index()
)

display(missing_summary)
display(missing_by_target)
```

**Análisis y decisión sobre valores faltantes**

De acuerdo con los resultados anteriores, el dataset no presenta valores faltantes ni filas incompletas. Esto significa que la integridad del conjunto no se ve comprometida por ausencia de datos y, por tanto, no fue necesario aplicar estrategias de imputación, relleno o eliminación de registros.

También se verificó la distribución de faltantes respecto a la variable objetivo `Diabetes_binary`, encontrando que ninguna de sus clases presenta vacíos de información. En consecuencia, la representatividad del problema binario balanceado se mantiene intacta y no hay riesgo de introducir sesgos derivados de imputaciones artificiales en esta etapa.


### **4.2. Valores duplicados**
---

Posteriormente se evalúa la presencia de registros duplicados exactos. Dado que el dataset no contiene un identificador único de paciente, es necesario distinguir entre un posible mismo paciente y un mismo perfil observado: la igualdad exacta entre filas solo permite diagnosticar perfiles repetidos, no confirmar identidad clínica individual.

```
duplicate_mask = diabetes_df.duplicated(keep="first")
duplicate_profile_mask = diabetes_df.duplicated(keep=False)
duplicated_rows = int(duplicate_mask.sum())
rows_in_duplicate_profiles = int(duplicate_profile_mask.sum())

duplicate_profile_summary = (
    diabetes_df.groupby(diabetes_df.columns.tolist(), dropna=False, as_index=False)
    .size()
    .rename(columns={"size": "Frecuencia"})
    .query("Frecuencia > 1")
    .sort_values("Frecuencia", ascending=False)
    .reset_index(drop=True)
)

duplicate_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros iniciales",
            "Duplicados exactos adicionales",
            "Registros involucrados en perfiles duplicados",
            "Porcentaje de duplicados exactos adicionales",
            "Perfiles exactos únicos con frecuencia > 1",
            "Registros conservados para modelado",
        ],
        "Valor": [
            len(diabetes_df),
            duplicated_rows,
            rows_in_duplicate_profiles,
            round((duplicated_rows / len(diabetes_df)) * 100, 2),
            len(duplicate_profile_summary),
            len(diabetes_df),
        ],
    }
)

duplicate_class_distribution = (
    diabetes_df.loc[duplicate_profile_mask, "Diabetes_binary"]
    .astype(int)
    .value_counts()
    .sort_index()
    .rename_axis("Diabetes_binary")
    .reset_index(name="Registros en perfiles duplicados")
)

clean_df = diabetes_df.copy()

display(duplicate_summary)
display(duplicate_class_distribution)
display(duplicate_profile_summary.head())
```

**Análisis y decisión sobre duplicados**

La inspección muestra la existencia de **23.899 registros duplicados exactos adicionales**, es decir, filas que repiten exactamente un perfil ya observado. Sin embargo, este conjunto de datos no incluye un identificador único de paciente, por lo que la coincidencia total entre variables no basta para afirmar que se trate del mismo individuo; también puede representar pacientes distintos con el mismo perfil clínico y sociodemográfico.

Por esta razón, los duplicados exactos no se eliminan automáticamente en la fase de limpieza. En su lugar, se conservan en el dataset y se controla el riesgo metodológico más importante: que perfiles idénticos queden simultáneamente en entrenamiento y prueba. Para evitar fuga de información y métricas artificialmente altas, el particionado supervisado se realiza por grupos de filas exactamente iguales, de modo que cada perfil repetido quede completo en `train` o completo en `test`, pero nunca en ambos conjuntos.


### **4.3. Valores atípicos**
---

Con el diagnóstico de perfiles duplicados ya establecido, se revisan los valores atípicos en las variables numéricas más sensibles desde el punto de vista clínico.

```
outlier_columns_clean = [col for col in ["BMI", "MentHlth", "PhysHlth"] if col in clean_df.columns]
outlier_rows_clean = []

for column in outlier_columns_clean:
    q1 = clean_df[column].quantile(0.25)
    q3 = clean_df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_count = int(((clean_df[column] < lower_bound) | (clean_df[column] > upper_bound)).sum())

    outlier_rows_clean.append(
        {
            "Variable": column,
            "Límite inferior IQR": round(lower_bound, 2),
            "Límite superior IQR": round(upper_bound, 2),
            "Posibles atípicos": outlier_count,
            "Porcentaje sobre clean_df": round((outlier_count / len(clean_df)) * 100, 2),
            "Mínimo observado": float(clean_df[column].min()),
            "Máximo observado": float(clean_df[column].max()),
        }
    )

clean_outlier_summary = pd.DataFrame(outlier_rows_clean)
display(clean_outlier_summary)
```

**Análisis y decisión sobre valores atípicos**

Los valores atípicos se identificaron mediante el criterio del rango intercuartílico (IQR), aplicado sobre `BMI`, `MentHlth` y `PhysHlth`, por ser variables donde los extremos pueden alterar la distribución de manera visible. Sin embargo, en este proyecto no se interpretan automáticamente como errores de medición.

Dado que se trata de un contexto clínico, los valores extremos pueden representar pacientes con condiciones de salud más severas, mayor carga de enfermedad o perfiles menos frecuentes pero reales. Por esta razón, la decisión final fue **conservar los valores atípicos** y controlar su efecto posteriormente mediante estandarización, en lugar de eliminarlos o transformarlos de forma agresiva.


### **4.4. Datos Inconsistentes**
---

En esta etapa se valida que las variables permanezcan dentro de los dominios esperados según la definición del dataset y se homogeneiza su representación para el modelado.

```
binary_columns = [
    "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke", "HeartDiseaseorAttack",
    "PhysActivity", "Fruits", "Veggies", "HvyAlcoholConsump", "AnyHealthcare",
    "NoDocbcCost", "DiffWalk", "Sex"
]

domain_rules = {
    "Diabetes_binary": {"allowed": [0, 1]},
    "GenHlth": {"allowed": [1, 2, 3, 4, 5]},
    "Age": {"allowed": list(range(1, 14))},
    "Education": {"allowed": [1, 2, 3, 4, 5, 6]},
    "Income": {"allowed": list(range(1, 9))},
    "MentHlth": {"range": (0, 30)},
    "PhysHlth": {"range": (0, 30)},
    "BMI": {"range": (1, None)},
}

for column in binary_columns:
    domain_rules[column] = {"allowed": [0, 1]}

inconsistent_row_mask = pd.Series(False, index=clean_df.index)
domain_check_rows = []

for column, rule in domain_rules.items():
    series = clean_df[column]
    if "allowed" in rule:
        invalid_mask = ~series.isin(rule["allowed"])
        criterio = f"Valores permitidos: {rule['allowed']}"
    else:
        lower, upper = rule["range"]
        invalid_mask = pd.Series(False, index=series.index)
        if lower is not None:
            invalid_mask |= series < lower
        if upper is not None:
            invalid_mask |= series > upper
        criterio = f"Rango esperado: {lower} a {upper if upper is not None else 'inf'}"

    inconsistent_row_mask |= invalid_mask
    domain_check_rows.append(
        {
            "Variable": column,
            "Criterio": criterio,
            "Registros fuera de dominio": int(invalid_mask.sum()),
        }
    )

domain_check_summary = pd.DataFrame(domain_check_rows)
integer_like_summary = pd.DataFrame(
    {
        "Variable": clean_df.columns,
        "Valores enteros codificados": [bool(np.allclose(clean_df[col], clean_df[col].round())) for col in clean_df.columns],
    }
)

clean_df = clean_df.round().astype(int)
consistent_df = clean_df.loc[~inconsistent_row_mask].reset_index(drop=True).copy()

display(domain_check_summary)
display(integer_like_summary)
display(pd.DataFrame({
    "Indicador": ["Registros con inconsistencias", "Registros después de validación de dominios"],
    "Valor": [int(inconsistent_row_mask.sum()), len(consistent_df)]
}))
```

**Análisis y decisión sobre inconsistencias**

La validación se realizó contrastando cada variable con su dominio esperado. En las variables binarias se comprobó la presencia exclusiva de valores `0` y `1`, incluida la variable objetivo `Diabetes_binary`; y en las variables ordinales y de conteo se revisaron sus rangos clínicamente plausibles.

Además, se observó que el archivo fue importado como `float64`, aunque sus valores están codificados como enteros. Para homogeneizar el formato y mejorar la coherencia del análisis, se convirtieron estas columnas a enteros antes de la fase de modelado. Esta decisión no altera la información, pero sí deja el dataset con una representación más consistente respecto a su significado original.


### **4.5. Datos corruptos**
---

Como complemento de la validación anterior, se revisa si existen registros corruptos derivados de errores de lectura, coerción numérica o valores infinitos.

```
coerced_df = consistent_df.apply(pd.to_numeric, errors="coerce")
coercion_failures = int(((coerced_df.isna()) & (~consistent_df.isna())).sum().sum())
infinite_values = int(np.isinf(coerced_df.to_numpy()).sum())
corrupt_row_mask = coerced_df.isna().any(axis=1) | np.isinf(coerced_df.to_numpy()).any(axis=1)

corrupt_summary = pd.DataFrame(
    {
        "Indicador": [
            "Fallos de coerción numérica",
            "Valores infinitos",
            "Registros corruptos detectados",
        ],
        "Valor": [
            coercion_failures,
            infinite_values,
            int(corrupt_row_mask.sum()),
        ],
    }
)

prepared_df = consistent_df.loc[~corrupt_row_mask].reset_index(drop=True).copy()

display(corrupt_summary)
display(pd.DataFrame({
    "Indicador": ["Registros listos después del control de corrupción"],
    "Valor": [len(prepared_df)]
}))
```

**Análisis y decisión sobre datos corruptos**

La búsqueda de datos corruptos se enfocó en tres señales: valores no convertibles a formato numérico, valores infinitos y registros que, después de la validación de dominios, siguieran mostrando comportamientos incompatibles con la estructura del dataset. Esta revisión es especialmente importante antes de alimentar algoritmos como una red neuronal o `K-means`, que requieren matrices numéricas consistentes.

Con base en la inspección realizada, el conjunto no presenta evidencia de corrupción estructural. Por ello, no fue necesario eliminar registros adicionales por este motivo y el dataset preparado conserva únicamente las observaciones válidas y consistentes obtenidas tras la deduplicación.


### **4.6. Selección de datos**
---

Finalmente, se define qué variables se conservan para cada uno de los enfoques de modelado y se justifica su inclusión.

```
target_column = "Diabetes_binary"
binary_class_ids = [0, 1]
binary_class_labels = ["Sin diabetes (0)", "Prediabetes o diabetes (1)"]

feature_columns = [
    column for column in prepared_df.columns
    if column != target_column
]

binary_feature_columns = [column for column in binary_columns if column in feature_columns]
ordinal_feature_columns = [column for column in ["GenHlth", "Age", "Education", "Income"] if column in feature_columns]
count_feature_columns = [column for column in ["BMI", "MentHlth", "PhysHlth"] if column in feature_columns]

selection_summary = pd.DataFrame(
    [
        {"Grupo": "Objetivo supervisado", "Variables": target_column, "Uso supervisado": "Variable objetivo binaria balanceada", "Uso K-means": "Se excluye"},
        {"Grupo": "Binarias", "Variables": ", ".join(binary_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
        {"Grupo": "Ordinales", "Variables": ", ".join(ordinal_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
        {"Grupo": "Conteo / magnitud", "Variables": ", ".join(count_feature_columns), "Uso supervisado": "Se incluyen", "Uso K-means": "Se incluyen"},
    ]
)

selection_size_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros disponibles para modelado",
            "Variables predictoras",
            "Variables en el problema supervisado",
            "Clases del problema supervisado",
            "Variables en el problema no supervisado",
        ],
        "Valor": [
            len(prepared_df),
            len(feature_columns),
            len(feature_columns) + 1,
            len(binary_class_ids),
            len(feature_columns),
        ],
    }
)

display(selection_summary)
display(selection_size_summary)
```

**Análisis y decisión sobre selección de datos**

No se aplicó muestreo ni reducción del tamaño del conjunto, ya que el volumen disponible sigue siendo adecuado para el modelado y conserva variabilidad clínica relevante. Los duplicados exactos tampoco se eliminan por falta de un identificador de paciente que permita demostrar que corresponden al mismo individuo; en cambio, se controlan posteriormente mediante un particionado por grupos exactos para evitar fuga entre entrenamiento y prueba.

Para el enfoque supervisado se utiliza directamente `Diabetes_binary`, donde `0` indica ausencia de diabetes y `1` agrupa los casos de prediabetes o diabetes. Las demás variables se incluyen como predictoras. Para el enfoque no supervisado con `K-means`, se excluye la etiqueta objetivo y se utiliza únicamente el espacio de características, con el fin de explorar agrupamientos sin introducir la condición clínica en el proceso de clustering.


## **5. Dataset listo para modelado**
---

A continuación se generan dos salidas listas para la siguiente fase: una para clasificación supervisada binaria basada directamente en `Diabetes_binary` y otra para `K-means`.


```
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler

prepared_with_groups = prepared_df.copy()
prepared_with_groups["duplicate_group_id"] = (
    prepared_with_groups.groupby(prepared_df.columns.tolist(), dropna=False, sort=False).ngroup()
)

X_supervised = prepared_with_groups[feature_columns].copy()
y_supervised = prepared_with_groups[target_column].astype(int).copy()
duplicate_groups = prepared_with_groups["duplicate_group_id"].copy()

group_splitter = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
train_idx, test_idx = next(
    group_splitter.split(X_supervised, y_supervised, groups=duplicate_groups)
)

X_train = X_supervised.iloc[train_idx].copy()
X_test = X_supervised.iloc[test_idx].copy()
y_train = y_supervised.iloc[train_idx].copy()
y_test = y_supervised.iloc[test_idx].copy()

train_group_ids = set(duplicate_groups.iloc[train_idx].tolist())
test_group_ids = set(duplicate_groups.iloc[test_idx].tolist())
shared_duplicate_groups = len(train_group_ids & test_group_ids)

supervised_scaler = StandardScaler()
X_train_scaled = pd.DataFrame(
    supervised_scaler.fit_transform(X_train),
    columns=feature_columns,
    index=X_train.index,
)
X_test_scaled = pd.DataFrame(
    supervised_scaler.transform(X_test),
    columns=feature_columns,
    index=X_test.index,
)

X_unsupervised = prepared_df[feature_columns].copy()
kmeans_scaler = StandardScaler()
X_kmeans_scaled = pd.DataFrame(
    kmeans_scaler.fit_transform(X_unsupervised),
    columns=feature_columns,
    index=X_unsupervised.index,
)

group_split_summary = pd.DataFrame(
    {
        "Indicador": [
            "Registros listos para modelado",
            "Grupos exactos únicos",
            "Grupos con frecuencia mayor a 1",
            "Registros en train",
            "Registros en test",
            "Grupos compartidos entre train y test",
        ],
        "Valor": [
            len(prepared_with_groups),
            duplicate_groups.nunique(),
            int((duplicate_groups.value_counts() > 1).sum()),
            len(X_train),
            len(X_test),
            shared_duplicate_groups,
        ],
    }
)

model_ready_summary = pd.DataFrame(
    {
        "Dataset": [
            "Supervisado - X_train_scaled",
            "Supervisado - X_test_scaled",
            "Supervisado - y_train",
            "Supervisado - y_test",
            "No supervisado - X_kmeans_scaled",
        ],
        "Filas": [
            len(X_train_scaled),
            len(X_test_scaled),
            len(y_train),
            len(y_test),
            len(X_kmeans_scaled),
        ],
        "Columnas": [
            X_train_scaled.shape[1],
            X_test_scaled.shape[1],
            1,
            1,
            X_kmeans_scaled.shape[1],
        ],
    }
)

class_distribution_summary = pd.concat(
    [
        y_supervised.value_counts().sort_index().rename("Total"),
        y_train.value_counts().sort_index().rename("Train"),
        y_test.value_counts().sort_index().rename("Test"),
    ],
    axis=1,
).fillna(0).astype(int)
class_distribution_summary.index.name = target_column
class_distribution_summary["Clase"] = class_distribution_summary.index.map(
    dict(zip(binary_class_ids, binary_class_labels))
)
class_distribution_summary = class_distribution_summary.reset_index()[
    [target_column, "Clase", "Total", "Train", "Test"]
]

display(group_split_summary)
display(model_ready_summary)
display(class_distribution_summary)
display(X_train_scaled.head())
display(X_kmeans_scaled.head())
```

**Resultado final de la preparación**

Después del proceso de limpieza, el conjunto queda listo para los dos enfoques definidos. En el caso supervisado, se trabaja directamente con `Diabetes_binary`, donde `0` representa a la población sin diabetes y `1` agrupa los casos de prediabetes o diabetes. Sobre esta etiqueta binaria se realiza un particionado estratificado por grupos de perfiles exactamente iguales, de manera que los registros duplicados permanezcan completos en entrenamiento o completos en prueba, pero nunca repartidos entre ambos conjuntos. Esto reduce fuga de información y evita inflar las métricas del modelo.

Posteriormente, las variables predictoras se escalan para facilitar el entrenamiento de los modelos de clasificación. Para el enfoque no supervisado, se prepara una matriz `X_kmeans_scaled` sin la etiqueta objetivo y con todas las características estandarizadas, de modo que `K-means` opere sobre variables comparables en escala.

```
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

import optuna
import random

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    accuracy_score, f1_score,
    roc_auc_score, balanced_accuracy_score,
    confusion_matrix, classification_report
)

import matplotlib.pyplot as plt
import seaborn as sns

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```
X_train_np = X_train_scaled.to_numpy()
X_test_np = X_test_scaled.to_numpy()

y_train_np = y_train.to_numpy()
y_test_np = y_test.to_numpy()

groups_train_np = duplicate_groups.iloc[train_idx].to_numpy()
```

```
def create_loaders(X, y, batch_size):
    X_t = torch.tensor(X, dtype=torch.float32)
    y_t = torch.tensor(y, dtype=torch.float32)

    dataset = TensorDataset(X_t, y_t)
    return DataLoader(dataset, batch_size=batch_size, shuffle=True)
```

```
def create_model(trial, input_dim):
    n_layers = trial.suggest_int('n_layers', 1, 5)
    layers = []

    in_features = input_dim

    for i in range(n_layers):
        out_features = trial.suggest_int(f'n_units_l{i}', 8, 128)

        layers.append(nn.Linear(in_features, out_features))
        layers.append(nn.ReLU())

        dropout_rate = trial.suggest_float(f'dropout_l{i}', 0.0, 0.3)
        layers.append(nn.Dropout(dropout_rate))

        in_features = out_features

    layers.append(nn.Linear(in_features, 1))

    return nn.Sequential(*layers)
```

```
def train_model(model, train_loader, val_loader, optimizer, epochs=30, patience=5):
    criterion = nn.BCEWithLogitsLoss()

    best_bal_acc = 0
    patience_counter = 0

    history = {
        "train_loss": [],
        "val_loss": [],
        "bal_acc": []
    }

    for epoch in range(epochs):
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            optimizer.zero_grad()
            logits = model(X_batch).view(-1)
            loss = criterion(logits, y_batch)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        history["train_loss"].append(train_loss / len(train_loader))

        # VALIDACIÓN
        model.eval()
        val_preds, val_targets = [], []
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch = X_batch.to(device)
                y_batch = y_batch.to(device)

                logits = model(X_batch).view(-1)
                loss = criterion(logits, y_batch)
                val_loss += loss.item()

                probs = torch.sigmoid(logits).cpu().numpy()
                preds = (probs >= 0.5).astype(int)

                val_preds.extend(preds)
                val_targets.extend(y_batch.cpu().numpy())

        val_loss /= len(val_loader)
        bal_acc = balanced_accuracy_score(val_targets, val_preds)

        history["val_loss"].append(val_loss)
        history["bal_acc"].append(bal_acc)

        if bal_acc > best_bal_acc:
            best_bal_acc = bal_acc
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= patience:
            break

    return history, best_bal_acc
```

```
def objective(trial):
    lr = trial.suggest_float('lr', 1e-5, 1e-2, log=True)
    batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])
    optimizer_name = trial.suggest_categorical('optimizer', ['Adam', 'SGD'])

    gkf = StratifiedGroupKFold(n_splits=3, shuffle=True, random_state=SEED)

    scores = []

    for train_idx_cv, val_idx_cv in gkf.split(X_train_np, y_train_np, groups_train_np):

        X_tr, X_val = X_train_np[train_idx_cv], X_train_np[val_idx_cv]
        y_tr, y_val = y_train_np[train_idx_cv], y_train_np[val_idx_cv]

        train_loader = create_loaders(X_tr, y_tr, batch_size)
        val_loader = create_loaders(X_val, y_val, batch_size)

        model = create_model(trial, input_dim=X_train_np.shape[1]).to(device)
        optimizer = getattr(optim, optimizer_name)(model.parameters(), lr=lr)

        _, best_bal_acc = train_model(model, train_loader, val_loader, optimizer)

        scores.append(best_bal_acc)

    mean_score = np.mean(scores)
    trial.set_user_attr("cv_balanced_accuracy", scores)

    return mean_score
```

```
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=10)

print("Mejores Parámetros:", study.best_params)
```

```
params = study.best_params

train_loader = create_loaders(
    X_train_np,
    y_train_np,
    params["batch_size"]
)

model = create_model(study.best_trial, X_train_np.shape[1]).to(device)
optimizer = getattr(optim, params["optimizer"])(model.parameters(), lr=params["lr"])

train_model(model, train_loader, train_loader, optimizer)
```

Para determinar el número óptimo de clústeres, se implementará el Método del Codo junto con la métrica de Silhouette Score. Teniendo en cuenta que método del codo busca minimizar la varianza dentro de cada grupo sin sobresegmentar los datos, mientras que el Silhouette Score busca maximizar la separación entre clústeres. ```
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


rango_k = range(2, 8)
inercia = []
silhouette_vals = []



for k in rango_k:
    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)

    kmeans_temp.fit(X_train_scaled)

    inercia.append(kmeans_temp.inertia_)

    sil_score = silhouette_score(X_train_scaled, kmeans_temp.labels_, sample_size=10000, random_state=42)
    silhouette_vals.append(sil_score)

fig, ax1 = plt.subplots(figsize=(10, 5))

# Gráfica del Método del Codo Azul
color = 'tab:blue'
ax1.set_xlabel('Número de Clústeres (k)', fontsize=12)
ax1.set_ylabel('Codo (Distancias al cuadrado)', color=color, fontsize=12)
ax1.plot(rango_k, inercia, marker='o', color=color, linewidth=2, markersize=8)
ax1.tick_params(axis='y', labelcolor=color)

# el Silhouette Score Roja
ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Silhouette Score', color=color, fontsize=12)
ax2.plot(rango_k, silhouette_vals, marker='s', color=color, linewidth=2, linestyle='dashed', markersize=8)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Búsqueda del K Óptimo: Método del Codo y Silhouette Score', fontsize=14)
fig.tight_layout()
plt.grid(True, alpha=0.3)
plt.show()
```

Con el método del codo observamos que la inercia disminuye a medida que aumentamos los clústeres. El "codo" no es extremadamente marcado en este caso, pero empieza a suavizarse alrededor de k=3 y k=4. # **Gradient Boosting**
---

Por último vamos a implementar el modelo de gradient boosting el cual va calificando de acuerdo a los parametros las iteraciones buscando maximizar las coincidencias de los datos, puesto que el entrenamiento posterior se realiza sobre los errores, en nuestro caso tenemos lo siguiente

```
import optuna
from sklearn.pipeline import Pipeline
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report

def objective_gb(trial):
    # Parámetros para optimizar en Gradient Boosting
    learning_rate = trial.suggest_float('learning_rate', 0.01, 0.2, log=True)
    n_estimators = trial.suggest_int('n_estimators', 50, 200)
    max_depth = trial.suggest_int('max_depth', 3, 7)


    pipeline_gb = Pipeline([
        ('scaler', StandardScaler()),
        ('gb', GradientBoostingClassifier(
            learning_rate=learning_rate,
            n_estimators=n_estimators,
            max_depth=max_depth,
            random_state=42
        ))
    ])


    score = cross_val_score(pipeline_gb, X_train_scaled, y_train, cv=3, scoring='accuracy').mean()
    return score


study_gb = optuna.create_study(direction='maximize')
study_gb.optimize(objective_gb, n_trials=10)

print(f"\nMejores parámetros: {study_gb.best_params}")

# Entrenar el Modelo Final con los mejores parámetros
best_pipeline_gb = Pipeline([
    ('scaler', StandardScaler()),
    ('gb', GradientBoostingClassifier(**study_gb.best_params, random_state=42))
])

best_pipeline_gb.fit(X_train, y_train)

# 3. Predicción y Evaluación
y_pred_gb = best_pipeline_gb.predict(X_test)
print("\n--- Reporte de Clasificación (Gradient Boosting) ---")
print(classification_report(y_test, y_pred_gb))
```

Tras la optimización de hiperparámetros con Optuna, el modelo Gradient Boosting convergió hacia una arquitectura con 72 estimadores y profundidad máxima de 3.

Evaluando el conjunto de prueba, el modelo alcanzó un Accuracy global del 75%. Sin embargo, el hallazgo más destacable dado el contexto de salud pública es el Recall del 80% para la clase positiva (Diabetes/Pre-diabetes). Esto indica que el modelo es altamente sensible y logra identificar correctamente a 8 de cada 10 pacientes en riesgo. Aunque la precisión para esta clase es del 73% lo que implica que tenemos algunos falsos positivos lo cual aunque es bueno en un escenario real mostraria mayor gasto en recursos, ahora para la categoría de paciente sano tenemos que tiene una precisión del 78% y un recall del 71%. ```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

rango_k = range(2, 7)


sil_scores = []
db_scores = []


for k in rango_k:

    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans_temp.fit_predict(X_train_scaled)


    sil = silhouette_score(X_train_scaled, labels, sample_size=15000, random_state=42)
    db = davies_bouldin_score(X_train_scaled, labels)

    sil_scores.append(sil)
    db_scores.append(db)


tabla_comparativa = pd.DataFrame({
    'Número de Clústeres (k)': rango_k,
    'Coef. de Silueta (↑ Mejor)': sil_scores,
    'Índice Davies-Bouldin (↓ Mejor)': db_scores
})

print("\n--- Tabla de Evaluación de Agrupamiento ---")
display(tabla_comparativa.round(4))

fig, ax1 = plt.subplots(figsize=(10, 5))


color1 = 'tab:blue'
ax1.set_xlabel('Número de Clústeres (k)', fontsize=12)
ax1.set_ylabel('Silhouette Score', color=color1, fontsize=12)
ax1.plot(rango_k, sil_scores, marker='o', color=color1, linewidth=2, label='Silueta (↑ Mejor)')
ax1.tick_params(axis='y', labelcolor=color1)


ax2 = ax1.twinx()
color2 = 'tab:red'
ax2.set_ylabel('Índice Davies-Bouldin', color=color2, fontsize=12)
ax2.plot(rango_k, db_scores, marker='s', color=color2, linestyle='dashed', linewidth=2, label='Davies-Bouldin (↓ Mejor)')
ax2.tick_params(axis='y', labelcolor=color2)

plt.title('Comparación de Métricas de Agrupamiento (K-Means)', fontsize=14)
fig.tight_layout()
plt.grid(True, alpha=0.3)
plt.show()
```

### **1.4.1. Descripción**
---

La solución propuesta se estructura como un pipeline modular para el procesamiento de imágenes orientado a la detección automática de emociones faciales. El flujo inicia con un usuario, cliente o proceso de prueba que envía una imagen al componente de ingesta. Este componente se implementa con FastAPI y expone el endpoint `POST /upload`, encargado de recibir el archivo de imagen.

Una vez recibida la imagen, el sistema la almacena en Cloudinary, que actúa como repositorio cloud de objetos multimedia. Cloudinary retorna una URL y metadatos del recurso cargado. Posteriormente, MongoDB Atlas almacena la información asociada a la imagen, incluyendo identificador, nombre de archivo, URL, fecha de carga, estado del procesamiento y resultado emocional cuando exista.

El componente de procesamiento consulta las imágenes registradas o trabaja sobre lotes del dataset público. Para pruebas con mayor volumen, Dask permite paralelizar tareas de preparación e inferencia. El modelo preentrenado `abhilash88/face-emotion-detection` clasifica la emoción facial y el resultado se actualiza en MongoDB Atlas.

Finalmente, el dashboard construido con Plotly consulta los resultados y presenta métricas agregadas como distribución de emociones, conteo de imágenes procesadas y estados generales del flujo.

Los componentes principales de la solución son:

- Cliente o fuente de imágenes.
- Endpoint de ingesta con FastAPI.
- Almacenamiento cloud de imágenes en Cloudinary.
- Base de datos MongoDB Atlas para metadatos, estados y resultados.
- Procesamiento batch con Dask.
- Modelo preentrenado de Hugging Face para inferencia emocional.
- Dashboard analítico con Plotly en el notebook.

Como alternativa de diseño se consideró usar una base adicional para trazabilidad de eventos, como Cassandra. Sin embargo, para el alcance actual se descartó porque agregaría complejidad operativa sin una necesidad directa. MongoDB Atlas es suficiente para centralizar los datos requeridos en esta fase. También podría incorporarse un sistema de colas en una versión futura para desacoplar completamente la ingesta y el procesamiento, pero se deja fuera del alcance inicial para mantener el prototipo manejable.

```python
nest_asyncio.apply()

app = FastAPI(title="Detección de Emociones")

@app.post("/upload")
async def upload_image(file: UploadFile = File(...)):
    
    return {
        "filename": file.filename,
        "status": "Recibido exitosamente",
        "message": "Listo para enviarse al almacenamiento cloud y procesamiento Batch."
    }

# Arrancar el servidor en un hilo separado
def levantar_servidor():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

# Iniciamos el servidor en segundo plano
threading.Thread(target=levantar_servidor, daemon=True).start()
print("Servidor FastAPI corriendo en segundo plano en http://127.0.0.1:8000")

# --- Prueba automatizada ---
try:
    with open("test_image.jpg", "wb") as f:

    with open("test_image.jpg", "rb") as f:
        response = requests.post("http://127.0.0.1:8000/upload", files={"file": f})
    
    print("Respuesta del servidor FastAPI:")
    print(response.json())
    print("El componente de ingesta funciona correctamente.")
except Exception as e:
    print(f"Error al probar el endpoint: {e}")
```

Primero, se crea una función que revisa la imagen y la transforma en un formato válido. Si detecta que está en blanco y negro, tiene transparencias (RGBA) o cualquier otro formato inusual, convierte la imágen a un formato estándar de 3 canales de color (RGB). Esto evita que el modelo de Hugging Face colapse por recibir formatos inesperados. Además, el redimensionamiento toma la imagen a exactamente 224x224 píxeles.

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
TARGET_SIZE = (224, 224)

def preparar_imagen_para_modelo(ruta_img, target_size=TARGET_SIZE):
    """
    Abre una imagen, la estandariza a 3 canales (RGB) y la redimensiona.
    """
    img = Image.open(ruta_img)


    if img.mode != 'RGB':
        img = img.convert('RGB')


    img = img.resize(target_size)

    return img

muestra_prueba = imagenes_validas[:5]
imagenes_preparadas = []

fig, axes = plt.subplots(1, len(muestra_prueba), figsize=(15, 4))

for i, ruta in enumerate(muestra_prueba):
    img_lista = preparar_imagen_para_modelo(ruta)
    imagenes_preparadas.append(img_lista)

    axes[i].imshow(img_lista)
    axes[i].set_title(f"Modo: {img_lista.mode}\nTam: {img_lista.size}")
    axes[i].axis('off')

plt.show()
````

Por último, se empleó un modelo de detección de caras para recortar las imágenes y solo dejar el rostro, así eliminando pixeles y características que pueda generar ruido para el modelo de Hugging Face.

````python
import cv2
import matplotlib.pyplot as plt
import os

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

def detectar_y_recortar_rostro(ruta_img):

    img = cv2.imread(ruta_img)

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    rostros = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    if len(rostros) > 0:
        x, y, w, h = rostros[0]
        recorte = img_rgb[y:y+h, x:x+w]
        return img_rgb, recorte, True
    else:

        return img_rgb, img_rgb, False

fig, axes = plt.subplots(3, 2, figsize=(8, 10))

for i, ruta in enumerate(imagenes_validas[10:13]):
    img_orig, img_recortada, detectado = detectar_y_recortar_rostro(ruta)


    axes[i, 0].imshow(img_orig)
    axes[i, 0].set_title("Original")
    axes[i, 0].axis('off')

    axes[i, 1].imshow(img_recortada)
    titulo = "Rostro Recortado" if detectado else "NO SE DETECTÓ ROSTRO"
    axes[i, 1].set_title(titulo)
    axes[i, 1].axis('off')

plt.tight_layout()
plt.show()
````

````python
from pymongo import MongoClient
import pandas as pd
import plotly.express as px
from IPython.display import display

LABEL_MAP = {
    "LABEL_0": "angry",
    "LABEL_1": "disgust",
    "LABEL_2": "fear",
    "LABEL_3": "happy",
    "LABEL_4": "sad",
    "LABEL_5": "surprise",
    "LABEL_6": "neutral",
}

client = MongoClient(MONGODB_URI)
collection = client[MONGODB_DATABASE][MONGODB_COLLECTION]

query = {"metadata.cohort": COHORT_FILTER}
projection = {
    "source_id": 1,
    "top_emotion": 1,
    "predictions": 1,
    "created_at": 1,
    "metadata": 1,
    "model_id": 1,
    "model_provider": 1,
    "cloudinary.secure_url": 1
}

documents = list(
    collection.find(query, projection).sort("created_at", -1).limit(LIMIT)
)
client.close()

for document in documents:
    document["prediction_id"] = str(document.pop("_id"))

print(f"Documentos recuperados para cohort={COHORT_FILTER!r}: {len(documents)}")
````

````python
if not documents:
    print("No se encontraron documentos para la cohorte configurada. Ajusta COHORT_FILTER o sube más imágenes.")
    df_predictions = pd.DataFrame()
    df_scores = pd.DataFrame()
else:
    df_predictions = pd.DataFrame(documents)
    df_predictions["created_at"] = pd.to_datetime(df_predictions["created_at"], errors="coerce")
    df_predictions["filename"] = df_predictions["metadata"].apply(lambda x: x.get("filename") if isinstance(x, dict) else None)
    df_predictions["source"] = df_predictions["metadata"].apply(lambda x: x.get("source") if isinstance(x, dict) else None)
    df_predictions["cohort"] = df_predictions["metadata"].apply(lambda x: x.get("cohort") if isinstance(x, dict) else None)
    df_predictions["top_emotion_name"] = df_predictions["top_emotion"].map(LABEL_MAP).fillna(df_predictions["top_emotion"])

    rows = []
    for document in documents:
        for score in document.get("predictions", []):
            rows.append({
                "prediction_id": document.get("prediction_id"),
                "source_id": document.get("source_id"),
                "top_emotion": document.get("top_emotion"),
                "top_emotion_name": LABEL_MAP.get(document.get("top_emotion"), document.get("top_emotion")),
                "label": score.get("label"),
                "label_name": LABEL_MAP.get(score.get("label"), score.get("label")),
                "score": score.get("score"),
                "created_at": document.get("created_at"),
                "filename": document.get("metadata", {}).get("filename"),
                "source": document.get("metadata", {}).get("source"),
                "cohort": document.get("metadata", {}).get("cohort"),
            })

    df_scores = pd.DataFrame(rows)
    if not df_scores.empty:
        df_scores["created_at"] = pd.to_datetime(df_scores["created_at"], errors="coerce")

    display(df_predictions[["prediction_id", "source_id", "top_emotion", "top_emotion_name", "created_at", "filename"]].head())
    if not df_scores.empty:
        display(df_scores[["prediction_id", "top_emotion_name", "label_name", "score", "created_at"]].head())
````

````python
if documents:
    summary = (
        df_predictions.groupby("top_emotion_name", dropna=False)
        .size()
        .reset_index(name="count")
        .sort_values("count", ascending=False)
    )
    display(summary)

    fig_top = px.bar(
        summary,
        x="top_emotion_name",
        y="count",
        title="Frecuencia de emoción dominante en la cohorte humans",
        text="count"
    )
    fig_top.update_layout(xaxis_title="Emoción dominante", yaxis_title="Número de imágenes")
    fig_top.show()
````

````python
if not df_scores.empty:
    fig_scores = px.box(
        df_scores,
        x="label_name",
        y="score",
        color="label_name",
        title="Distribución de scores por emoción"
    )
    fig_scores.update_layout(xaxis_title="Emoción", yaxis_title="Score")
    fig_scores.show()
else:
    print("No hay scores individuales para graficar.")
````

````python
if documents:
    timeline = (
        df_predictions.assign(date=df_predictions["created_at"].dt.floor("min"))
        .groupby(["date", "top_emotion_name"], dropna=False)
        .size()
        .reset_index(name="count")
        .sort_values("date")
    )

    fig_timeline = px.line(
        timeline,
        x="date",
        y="count",
        color="top_emotion_name",
        markers=True,
        title="Evolución temporal de las emociones dominantes"
    )
    fig_timeline.update_layout(xaxis_title="Fecha de procesamiento", yaxis_title="Cantidad de predicciones")
    fig_timeline.show()
````

> **Nota:** en esta sección se usa el mapeo `LABEL_0`→`angry`, `LABEL_1`→`disgust`, `LABEL_2`→`fear`, `LABEL_3`→`happy`, `LABEL_4`→`sad`, `LABEL_5`→`surprise` y `LABEL_6`→`neutral`, según la documentación y la app pública del modelo `abhilash88/face-emotion-detection`.
